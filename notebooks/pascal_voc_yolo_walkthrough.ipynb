# YOLOv1-Style Detection on Pascal VOC

This walkthrough packages a grid-based object detector, decomposed loss, and class-aware non-max suppression.

## Detection representation

Each 7×7 cell predicts two `(confidence, x, y, width, height)` boxes and 20 class probabilities.

```python
import torch
from pascal_voc_yolo.data import encode_voc_target
from pascal_voc_yolo.loss import yolo_loss

target = encode_voc_target(torch.tensor([[20., 30., 100., 150.]]), torch.tensor([14]), 224, 224)
prediction = torch.zeros_like(target).unsqueeze(0)
losses = yolo_loss(prediction, target.unsqueeze(0))
losses
```

## Model and postprocessing

A ResNet-50 feature extractor feeds a compact YOLO head. Decoded boxes are filtered with deterministic, class-aware NMS.
